# Virtual Environment Management in Python

## Why Use Virtual Environments?

Creating a virtual environment for each project ensures:

- Package versions remain consistent, avoiding accidental updates that could break the code.

- Your project dependencies are isolated, preventing conflicts between different projects.

It is recommended to store the virtual environment within the project folder for better organization.

<br>

---

## Setting Up PowerShell - (This is only required for Windows)

To properly work with virtual environments in PowerShell:

1. Run PowerShell as an administrator.

2. Execute the following command to enable script execution:
   ```powershell
   Set-ExecutionPolicy RemoteSigned
   ```
   
   - When prompted, type Y and press Enter. 
   
   This step ensures you can activate virtual environments later.

The `Set-ExecutionPolicy RemoteSigned` command is used in PowerShell to change the script execution policy on your system. By setting it to RemoteSigned, you're allowing scripts to run on your machine if they are locally created, but requiring that scripts downloaded from the internet (or from a remote source) are signed by a trusted publisher.

## Creating New Virtual Environment

1. Navigate to the desired folder in PowerShell/Zsh.

2. Create a virtual environment using the command: `python -m venv project_env`

    Replace `project_env` with your preferred virtual environment name.

When you create a new virtual environment in Python, the default Python version used will typically be the one that is active on your system. For example, if you have Python 3.9 installed and it's the default version on your system, a new virtual environment will use Python 3.9 by default.

However, if you want to specify a different Python version for the virtual environment, you can do so by providing the path to the desired Python interpreter when creating the virtual environment. For example: `python3.13 -m venv myenv`

## Checking Python Version in the Virtual Environment

- To verify the Python version associated with the virtual environment: `python --version`

<br>

## Activate Virtual Environment

#### Windows:

- Navigate into the virtual environment folder: `cd project_env\Scripts`

- Run the activation script: `Activate.ps1`

- After activation, the PowerShell prompt will indicate the virtual environment is active.


#### Mac:

- Navigate into the virtual environment folder: `cd project_env/bin`

- Run the activation script: `source activate`

- After activation, the PowerShell prompt will indicate the virtual environment is active.

<br>

---

## Decativate Virtual Environment

- Simply write `deactivate` into the terminal.

<br>

---

## Deleting a Virtual Environment

### Windows:

To delete a virtual environment, use: `rmdir project_env /s`. The `/s` flag ensures that non-empty directories are removed.

### Mac:

To delete a virtual environment, use: `rm -rf project_env`. 

**Explanation:**

- `rm`: Removes files and directories.

- `-r`: Recursively removes all contents inside the directory.

- `-f`: Forces deletion without prompting for confirmation.

## Managing Packages

Viewing Installed Packages: `pip list`

Installing a Package:

- Latest version: `pip install [package_name]`

- Specific version: `pip install [package_name]==[version_number]`, for example: `pip install numpy==1.22.4`

- To uninstall a specific package: `pip uninstall [package_name]`

- Check version: `pip show package_name`

- Exporting Installed Packages to a File: `pip freeze > requirements.txt`

- Installing Packages from `requirements.txt`: `pip install -r requirements.txt`

When a virtual environment is active, all pip and python commands apply only to that environment.

<br>

## Do You Need to Add New Environments to the PATH?

No, you do not need to add virtual environments to the system PATH manually. When you activate a virtual environment, PowerShell temporarily adjusts the PATH to point to the virtual environment's Scripts folder. This ensures that the correct Python interpreter and packages are used while the environment is active.
