# What is Poetry?

Poetry is a tool designed to help you manage Python projects, including dependencies, packaging, and virtual environments. 

It aims to streamline development by offering a centralized configuration file and an intuitive CLI (Command Line Interface) for common tasks.

<br>

---

## How to use Poetry?

- Install Poetry: `pip install poetry`

- Install Poetry shell plugin: `pip install poetry-plugin-shell`

- Create a New Project: `poetry new my_project`

    This creates a folder structure and basic configuration files like pyproject.toml.

---

In order to add dependencies / install / activate the environment, we need the poetry.lock and pyproject.toml in the folder. (What are those?)


pyproject.toml: This is the main configuration file where Poetry stores project metadata, including project name, version, description, and dependencies. This file is required to manage your project's configuration.
poetry.lock: This file is automatically generated when you add dependencies. It locks the specific versions of dependencies to ensure that everyone working on the project uses the same versions, even if they’re using different machines. It guarantees reproducibility across environments.


- Add dependencies using: `poetry add [package_name]`

    Poetry automatically resolves and locks compatible versions in poetry.lock

- Remove dependencies: `poetry remove [package_name]`

- Install dependencies for your project: `poetry install`

- Activate environment: `poetry shell`

- Deactivate environment: `exit`


View the Environment Information:
You can check details about the current virtual environment with: `poetry env info`

Run a Command in the Environment:
Instead of activating the environment, you can run commands directly within the environment using: `poetry run <command>`

Example: To run a Python script:

`poetry run python script.py`

What Happens When You Run poetry update?
Updates Dependencies:
Poetry will update all the dependencies in the pyproject.toml file that have new versions available that still satisfy the version constraints you’ve set.
Modifies the poetry.lock file:
The poetry.lock file will be updated to reflect the exact versions of the dependencies that were installed. This ensures that everyone working on the project will have the same versions installed, ensuring consistency across different environments.
Automatic Dependency Resolution:
Poetry resolves any dependency conflicts automatically and ensures that the new versions of dependencies are compatible with each other and with your project.
When to Use poetry update:
When you want to get the latest versions of the dependencies while respecting the version constraints you've set.
If you want to ensure that the dependencies in your poetry.lock file are up to date, especially when there's a new patch or minor release that could include important fixes or features.

Example Usage:
Updating All Dependencies: Run this command to update all of the dependencies in the project:

poetry update

Updating a Specific Dependency: You can also specify a particular package if you only want to update that one package:

poetry update requests

This will update the requests package to the latest version that satisfies the version constraint in pyproject.toml and modify the poetry.lock file.


Be Careful:
Breaking Changes: If a new version of a dependency introduces breaking changes, Poetry will respect the version constraints set in your pyproject.toml. However, if you don't have strict version constraints, running poetry update could potentially update a package to a new major version that introduces breaking changes.
To avoid surprises, it’s a good idea to pin your dependencies with more specific version constraints (e.g., requests = "^2.25.1" rather than requests = "*") and review changes when updating.
