# How to modify the `PATH` Variable?

## System-wide configuration

1. Open Terminal: Launch the Terminal application on your Mac.

2. Run the Command: `sudo nano /etc/paths`

    This command opens the `/etc/paths` file in the nano text editor with superuser (admin) privileges. The `/etc/paths` file contains a list of directories that are added to your `PATH` variable system-wide.

3. Enter Your Password: When prompted, enter your Mac's administrator password. It won't show the characters as you type for security reasons.

4. Edit the File:

    - Use the arrow keys to scroll to the bottom of the file.
    
    - Add the full path of the directory you want to include in the PATH. For example: `/usr/local/bin`

    Ensure that each path occupies a separate line.

5. Exit and Save:

    - Press Control-X to exit the nano editor.

    - Press Y to confirm saving changes.

    - Press Enter to overwrite the file.

## User-Level Changes

If you don't want to modify the system-wide PATH (or don't have admin access), you can modify your user-specific PATH by editing your shell configuration file (`~/.zshrc`, `~/.zshenv` or `~/.zprofile` if we use login shell)

1. Open the appropriate file in a text editor, e.g.:

    `nano ~/.zshrc`

2. Add the following line at the end of the file:

    `export PATH="/path/to/your/directory:$PATH"`

3. Save and close the file.

4. Reload the file: `source ~/.zshrc`
