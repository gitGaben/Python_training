# Understanding the `PATH` Environment Variable

The `PATH` is an environment variable that tells the shell which directories to search for executable files when you type a command. It allows you to run programs without providing their full file paths.

<br>

---

## How `PATH` Works

When you enter a command (e.g., `ls`, `python`, `git`), the shell searches through the directories listed in the `PATH` variable sequentially to locate the executable. The shell uses the first matching executable it finds. If the command is not found in any of the specified directories, the shell returns an error.

<br>

---

## Example

Even basic commands like `ls` are executable programs. For example, on macOS, the `ls` executable resides in `/bin`. To verify its location, you can use the `which ls` command, which will return `/bin/ls`.

When you type `ls` in the Terminal, the shell uses the `PATH` variable to locate the executable in directories like `/usr/bin`, `/bin`, etc., in the order they are listed.

**Fun fact:** You can also execute the `ls` command directly by specifying its full path: `/bin/ls`.

<br>

---

## Viewing the Current `PATH`

To see the current `PATH` value, run:

```bash
echo $PATH
```

The output will look something like this:

```bash
/usr/local/bin:/usr/bin:/bin:/usr/sbin:/sbin
```

Each directory in the `PATH` is separated by a colon (`:`):

- **/usr/local/bin**: User-installed programs or custom scripts (e.g., Homebrew).

- **/usr/bin**: System utilities provided by macOS (e.g., `python3`, `git`).

- **/bin**: Basic Unix tools (e.g., `ls`, `cat`, `rm`).

- **/usr/sbin** and **/sbin**: System administration commands (e.g., `ifconfig`, `shutdown`).

When you run a command, the shell checks each directory in the listed order until it finds the executable or reports "command not found."

<br>

## Components of the `PATH` Variable

The `PATH` variable is built from two main sources:

### 1. System-Wide Configuration Files

System-wide configuration files define default directories to be included in the `PATH`. On macOS, these defaults are stored in `/etc/paths`. They are used by an operating system to apply settings and configurations that affect all users on the system. These files are typically stored in specific directories that are restricted to users with administrative privileges. To view or edit them:

```bash
sudo nano /etc/paths
```

### 2. Dynamic Modifications During Shell Initialization

Other files and processes can dynamically modify `PATH` at runtime. These changes are not stored in files like `/etc/paths` but occur during shell initialization or command execution.

<br>

---

### Shell Startup Scripts

When a shell (e.g., Zsh) starts, it runs a series of initialization files that may modify `PATH`. Examples include:

- **User-specific scripts**:  

  `~/.zshrc`, `~/.zprofile`, `~/.zshenv`, `~/.bash_profile`

- **System-wide scripts**:  

  `/etc/zshrc`, `/etc/profile`, `/etc/zprofile`

For example, you might find a line in `~/.zshrc` like this:

```bash
export PATH="/custom/directory:$PATH"
```

When you add that line to your `~/.zshrc` file, it modifies the `PATH` environment variable every time a new Zsh shell session is started. The change will apply whenever you open a new terminal window or tab.

<br>

---

### Commands or Programs Modifying PATH

Some programs automatically modify `PATH` when installed or executed. Examples include:

- **Homebrew**:  

  Adds `/opt/homebrew/bin` to `PATH` during installation by appending a line to `~/.zprofile`:

  ```bash
  eval "$($(brew --prefix)/bin/brew shellenv)"
  ```

- **Environment Managers**:

  Tools like `pyenv`, `nvm`, or `rbenv` dynamically adjust `PATH` to prioritize specific versions of Python, Node.js, or Ruby.

<br>

---

### Temporary Modifications

You can temporarily modify `PATH` for the current session with a single command:

```bash
export PATH="/temporary/directory:$PATH"
```

## Zsh (Z Shell) Startup Scripts Overview

Zsh uses several initialization files to configure your shell environment, depending on the type of shell you are running. The main types of shells are:

- **Interactive Shell**: A shell where you interact with the system, like when you open a terminal window. It waits for commands and displays prompts.

- **Non-interactive Shell**: A shell that runs scripts or commands automatically without user interaction (e.g., when executing a script).

- **Login Shell**: A shell started as part of a login process (e.g., logging into the system via SSH, console, or a terminal session).

Each type of shell reads specific initialization files to configure the environment. Below is an overview of the files and when they are used:

#### `/etc/zshenv` (System-wide)
- **When it runs:** This script is executed for **all** Zsh sessions, whether interactive or non-interactive.

- **Purpose:** It is commonly used to set **global environment variables** that should apply to all users and shells. For example, system-wide variables like `PATH`, `EDITOR`, or custom variables.

- **Notes:** This file is sourced **before any other configuration files** to ensure environment variables are set early.

#### `~/.zshenv` (User-specific)

- **When it runs:** This script is executed for **every Zsh session**, regardless of whether it's interactive or non-interactive.

- **Purpose:** It is typically used to define **user-specific environment variables**, such as custom `PATH` entries or other user preferences that should apply universally across all Zsh sessions.

- **Notes:** Unlike `/etc/zshenv`, this file affects only the **current user**.

#### `/etc/zshrc` and `~/.zshrc` (Interactive Shells)

- **When they run:** These files are read when starting an **interactive shell**, meaning when you open a terminal or a new shell session that expects user interaction.

- **Purpose:** They are used to configure settings like **aliases**, **functions**, and other **interactive behaviors** of the shell (such as modifying the prompt, enabling autocompletion, etc.).

#### `/etc/zprofile` and `~/.zprofile` (Login Shells)

- **When they run:** These files are sourced during **login shell initialization**, which happens when you log into the system (e.g., via SSH or at the console).

- **Purpose:** They are used to set **environment variables** (e.g., `PATH`) and configure other settings needed during the **login** process.

#### `/etc/zlogin` and `~/.zlogin` (Post-Login Shells)

- **When they run:** These files are executed **after `zprofile`** for login shells.

- **Purpose:** These files are typically used for tasks that should be done **after the login process** has been completed, such as launching background processes or configuring session-specific settings.

#### `~/.zlogout` (Logout Tasks)

- **When it runs:** This script is executed when a Zsh session ends, typically when you exit the shell or log out of a terminal session.

- **Purpose:** It is used to perform any **cleanup tasks** such as clearing temporary files, saving history, or logging out processes that should run when the shell session terminates.

<br>

To check the contents of each startup script, simply use the cat command. For example: `cat ~/.zprofile`

---

## Key Points to Remember

### Order of Execution for Login Shells:

- `/etc/zshenv` → `~/.zshenv` → `/etc/zprofile` → `~/.zprofile` → `/etc/zlogin` → `~/.zlogin`

### Order for Interactive Non-Login Shells:

- `/etc/zshenv` → `~/.zshenv` → `/etc/zshrc` → `~/.zshrc`

### File Responsibilities:

- Use `.zshenv` for environment variables.

- Use `.zshrc` for interactive shell configurations.

- Use `.zprofile` and `.zlogin` for login-specific tasks.

### Priorities:

- If both system-wide (`/etc/...`) and user-specific (`~/...`) versions of a file exist, the user-specific file overrides the system-wide file. This applies to all cases, ensuring user preferences take precedence.

<br>

## macOS Terminal Initialization

By default, macOS terminal emulators (like Terminal.app or iTerm2) are configured to start login shells for each new terminal session. This is to ensure that environment variables and settings from files like `~/.zprofile` are loaded.

### What Happens During Initialization

When the terminal starts:

1. It runs as a login shell.

2. Zsh sources these files in order:

   - `/etc/zshenv` → `~/.zshenv` → `/etc/zprofile` → `~/.zprofile` → `/etc/zlogin` → `~/.zlogin`.

3. Since it’s also interactive, it will then source:

   - `/etc/zshrc` → `~/.zshrc`.
   
This combination ensures both login and interactive settings are loaded.

<br>

---

### The Difference Between Adding the PATH Variable to `~/.zshenv` vs. `~/.zshrc`:

The difference between adding the PATH variable (or any other environment variable) to `~/.zshenv` vs. `~/.zshrc` lies in when and how these files are sourced during the Zsh shell startup process.

Here’s a breakdown:

#### 1. Adding PATH to `~/.zshenv`

- **When it’s sourced:** `~/.zshenv` is sourced for every Zsh session, whether it's interactive or non-interactive.

- **Purpose:** This file is intended for environment variables and configurations that should be available for all Zsh sessions, regardless of whether they are interactive or non-interactive.

- **Best use case for PATH:** If you want the PATH to be set in all Zsh sessions, including those invoked by scripts or background processes (non-interactive shells), `~/.zshenv` is the correct place.

#### 2. Adding PATH to `~/.zshrc`

- **When it’s sourced:** `~/.zshrc` is sourced only for interactive shells. That means this file is read when you open a terminal or start an interactive shell session.

- **Purpose:** This file is mainly used for interactive shell settings like aliases, functions, and customizations that affect the behavior of the terminal during a session.

- **Best use case for PATH:** If you want the PATH variable to be adjusted only when you’re interacting with a terminal (for example, customizing the PATH for interactive tools like git or vim), then `~/.zshrc` would be a better place.

#### Key Differences:

- **Scope:**

    - `~/.zshenv` affects all Zsh sessions (interactive and non-interactive).

    - `~/.zshrc` only affects interactive Zsh sessions (those where you are working directly in the terminal).
  
- **When to use:**

    - Put PATH in `~/.zshenv` if you want it set for all types of sessions (including when scripts or other non-interactive shells are run).

    - Put PATH in `~/.zshrc` if you want it to only apply to interactive shell sessions (when you’re working directly in the terminal).

#### In Practice:

- For global environment variables that should be available everywhere (like PATH), it’s generally safer to put them in `~/.zshenv`.

- For terminal-specific configurations (like interactive commands or prompt changes), use `~/.zshrc`.

### Example from my `~/.zprofile`:

```bash
kristongabor@Kriston-MacBook-Pro ~ % cat ~/.zprofile

eval "$(/opt/homebrew/bin/brew shellenv)"

# Setting PATH for Python 3.11

# The original version is saved in .zprofile.pysave

PATH="/Library/Frameworks/Python.framework/Versions/3.11/bin:${PATH}"`

export PATH
```

<br>

### 1. Homebrew Shell Environment Initialization

#### What it does:

When you run the command `eval "$(/opt/homebrew/bin/brew shellenv)"`, it sets up your terminal environment to work with Homebrew. This includes modifying system environment variables, such as the `PATH`, to include the locations where Homebrew stores its executables and other binaries. This ensures that you can run Homebrew-managed commands and software from anywhere in your terminal without needing to specify their full paths.

#### Effect on `PATH`:

The command adds the following directories to your system's `PATH` variable:

- `/opt/homebrew/bin`

- `/opt/homebrew/sbin`

By doing this, it makes sure that binaries installed via Homebrew can be accessed globally in the terminal, allowing you to run Homebrew-managed software without any additional setup.


### 2. Setting PATH for Python 3.11

#### What it does:

The comment 

```bash
# Setting PATH for Python 3.11
```
is there to indicate that the following line is responsible for configuring the Python environment.

This line:

```bash
PATH="/Library/Frameworks/Python.framework/Versions/3.11/bin:${PATH}"
```
modifies the PATH by prepending (/Library/Frameworks/Python.framework/Versions/3.11/bin) to the front of your existing PATH (i.e., ${PATH}). It ensures that Python 3.11 is prioritized in the system path, meaning that if there are multiple Python versions installed, the one in /Library/Frameworks/Python.framework/Versions/3.11/bin will be the one used by default when you run python3 or pip3.

#### Effect on `PATH`:

This makes sure that the Python binaries from Python 3.11 are found first when you run Python commands in the terminal. The path /Library/Frameworks/Python.framework/Versions/3.11/bin is now part of your PATH.
